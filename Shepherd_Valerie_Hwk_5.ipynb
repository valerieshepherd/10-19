Explain why decision trees in prescriptive analytics (DADT) are not predictive models.

Decision trees in DADT are not predictive models because they focus on evaluating alternative decisions and their possible outcomes rather than primarily predicting an unknown outcome from historical data. They use probabilities, payoffs, and decision alternatives to identify the best decision.

A company is choosing between two investment locations: City Center and Suburb. Each location has three development options (No Development, Modest, Major), with associated prior probabilities and expected annual earnings. The investment horizon is 10 years, and the discount rate is 6%.

```
import pandas as pd

data = {
    "Location": ["City Center"] * 3 + ["Suburb"] * 3,
    "Development": ["No Dev", "Modest", "Major"] * 2,
    "Prior": [0.2, 0.3, 0.5, 0.3, 0.4, 0.3],
    "Annual_Earnings": [1.0, 1.4, 2.0, 1.1, 1.6, 1.9],
}

discount_rate = 0.06
term = 10

```

## Exercise 7(a) (5 points)

Calculate the present value of earnings for each scenario.

````python
import pandas as pd

data = {
    "Location": ["City Center"] * 3 + ["Suburb"] * 3,
    "Development": ["No Dev", "Modest", "Major"] * 2,
    "Prior": [0.2, 0.3, 0.5, 0.3, 0.4, 0.3],
    "Annual_Earnings": [1.0, 1.4, 2.0, 1.1, 1.6, 1.9]
}

discount_rate = 0.06
term = 10

df = pd.DataFrame(data)

annuity_factor = (1 - (1 + discount_rate) ** (-term)) / discount_rate

df["Present_Value"] = df["Annual_Earnings"] * annuity_factor

print(df.round(4))
````

## Exercise 7(b) (3 points)

Compute the expected present value per development type.

````python
df["Expected_PV"] = df["Prior"] * df["Present_Value"]

print(df[["Location", "Development", "Expected_PV"]].round(4))
````

## Exercise 7(c) (3 points)

Compute the total expected present value per location.

````python
location_values = df.groupby("Location")["Expected_PV"].sum()

print(location_values.round(4))
````

## Exercise 7(d) (3 points)

Recommend the better investment based on expected returns.

````python
print("Recommended investment:", location_values.idxmax())
print("Expected present value:", round(location_values.max(), 4))
````

# Exercise 8 (8 points)

A subscription company models its installed base of subscribers with the first-order difference equation

$$Y_t = \alpha + \beta Y_{t-1} + \gamma X_t,$$

where $Y_t$ is the number of subscribers (in thousands) in month $t$, $\beta$ is the retention rate, and $X_t$ is marketing spend (in thousands of dollars). Use the following inputs:

```
import numpy as np

y0 = 1000                      # starting subscribers (thousands)
alpha, beta, gamma = 50, 0.85, 0.8
x_base = np.repeat(200.0, 24)  # steady spend for 24 months
```

Consider a what-if scenario in which marketing spend is frozen (set to 0) in months 7 through 12 (i.e., positions 6 to 11 of the array, using zero-based indexing) and then returns to 200.

## Exercise 8(a) (3 points)

Write a function that iterates the difference equation forward, and use it to simulate the baseline path and the spending-freeze path.

````python
import numpy as np

y0 = 1000
alpha, beta, gamma = 50, 0.85, 0.8
x_base = np.repeat(200.0, 24)

x_freeze = x_base.copy()
x_freeze[6:12] = 0

def simulate(y0, x, alpha, beta, gamma):
    y = np.zeros(len(x) + 1)
    y[0] = y0

    for t in range(len(x)):
        y[t + 1] = alpha + beta * y[t] + gamma * x[t]

    return y

baseline = simulate(y0, x_base, alpha, beta, gamma)
freeze = simulate(y0, x_freeze, alpha, beta, gamma)

print("Baseline path:", baseline)
print("Freeze path:", freeze)
````

## Exercise 8(b) (3 points)

Compute the steady state of the baseline system, $Y^* = (\alpha + \gamma \bar{X})/(1-\beta)$, and compare it with the baseline value in month 24. Then compute the cumulative gap (baseline minus freeze, summed over all periods) to measure the cost of the freeze.

````python
steady_state = (alpha + gamma * 200) / (1 - beta)
cumulative_gap = np.sum(baseline[1:] - freeze[1:])

print("Steady state:", steady_state)
print("Baseline value in month 24:", baseline[24])
print("Cumulative gap:", cumulative_gap)
````

## Exercise 8(c) (2 points)

What is the largest gap between the baseline and the freeze path, and in which month does it occur? Explain in one or two sentences why the gap does not close immediately once spending resumes.

````python
gap = baseline[1:] - freeze[1:]
largest_gap = np.max(gap)
month = np.argmax(gap) + 1

print("Largest gap:", largest_gap)
print("Month:", month)
print(
    "The gap persists because lower subscriber counts during the freeze "
    "carry forward through the retention rate, even after spending resumes."
)
````

# Exercise 9 (12 points)

A retailer must decide how many units $Q$ of a seasonal product to stock. Each unit sells for \$60 and costs \$35. Units that are not sold are salvaged at \$12. Demand $D$ is uncertain and follows a normal distribution with mean 1,000 and standard deviation 250 (truncated at zero). The retailer is considering stocking $Q = 1{,}000$ units. Profit is

$$g(D) = 60\,\min(D, Q) - 35\,Q + 12\,\max(Q-D, 0).$$

```
import numpy as np

price, cost, salvage = 60, 35, 12
Q = 1000
mu, sigma = 1000, 250
N = 100_000
```

## Exercise 9(a) (2 points)

Write the profit as a function of demand and stocking level, and compute the profit when demand equals its average (1,000 units).

````python
import numpy as np

price, cost, salvage = 60, 35, 12
Q = 1000
mu, sigma = 1000, 250
N = 100_000

def profit(D, Q):
    return price * np.minimum(D, Q) - cost * Q + salvage * np.maximum(Q - D, 0)

print("Profit at average demand:", profit(1000, Q))
````

## Exercise 9(b) (4 points)

Using `np.random.default_rng(511)`, draw $N$ demand values from the normal distribution (clip negative draws to zero) and evaluate the profit for each. Report the mean simulated profit together with its Monte Carlo standard error and 95% confidence interval.

````python
rng = np.random.default_rng(511)
D = np.maximum(rng.normal(mu, sigma, N), 0)
profits = profit(D, Q)

mean_profit = np.mean(profits)
standard_error = np.std(profits, ddof=1) / np.sqrt(N)
ci_low = mean_profit - 1.96 * standard_error
ci_high = mean_profit + 1.96 * standard_error

print("Mean simulated profit:", mean_profit)
print("Monte Carlo standard error:", standard_error)
print("95% confidence interval:", (ci_low, ci_high))
````

## Exercise 9(c) (2 points)

Compare the profit at average demand from (a) with the expected profit from (b). Which one is larger and why? Relate your answer to the "flaw of averages" (Jensen's inequality).

````python
profit_at_average = profit(mu, Q)
expected_profit = np.mean(profits)

print("Profit at average demand:", profit_at_average)
print("Expected profit:", expected_profit)

if expected_profit > profit_at_average:
    print(
        "Expected profit is larger because the profit function is convex "
        "in demand, illustrating Jensen's inequality."
    )
else:
    print("Compare the simulated result with the profit at average demand.")
````

## Exercise 9(d) (4 points)

Compute the following risk metrics for the simulated profit: the probability of a loss ($P(\text{profit} < 0)$), the 5% Value at Risk (VaR), and the 5% Conditional Value at Risk (CVaR). Interpret the CVaR in one sentence.

````python
probability_of_loss = np.mean(profits < 0)
var_5 = np.percentile(profits, 5)
cvar_5 = np.mean(profits[profits <= var_5])

print("Probability of loss:", probability_of_loss)
print("5% VaR (profit threshold):", var_5)
print("5% CVaR (average bottom-tail profit):", cvar_5)
print(
    "The 5% CVaR represents the average profit in the worst 5% "
    "of simulated outcomes."
)
````

# Exercise 10 (12 points)

A firm is considering a four-year product launch. The fixed launch cost is \$2.5M and cash flows are discounted at 8%. The uncertain inputs and their distributions are:

| Input | Distribution |
|---|---|
| Year-1 demand (units) | Normal(mean = 40,000, sd = 6,000), clipped at 0 |
| Annual demand growth | Triangular(min = -0.03, mode = 0.03, max = 0.10) |
| Price per unit (\$) | Triangular(min = 55, mode = 60, max = 70) |
| Unit cost (\$) | Lognormal with median 38 and sigma = 0.12 |

Demand in year $t$ is $\text{demand}_1 (1+\text{growth})^{t-1}$, and the cash flow in year $t$ is $\text{demand}_t \times (\text{price} - \text{cost})$. The key performance metric is the NPV.

```
import numpy as np

N = 100_000
fixed, r, horizon = 2_500_000, 0.08, 4
```

## Exercise 10(a) (4 points)

Using `np.random.default_rng(2026)`, draw $N$ replications of each input (in the order listed above) and compute the NPV for each replication. Report the mean NPV with its 95% Monte Carlo confidence interval and the median NPV.

````python
import numpy as np

N = 100_000
fixed, r, horizon = 2_500_000, 0.08, 4

rng = np.random.default_rng(2026)

demand1 = np.maximum(rng.normal(40_000, 6_000, N), 0)
growth = rng.triangular(-0.03, 0.03, 0.10, N)
price = rng.triangular(55, 60, 70, N)
cost = rng.lognormal(np.log(38), 0.12, N)

npv = np.full(N, -fixed, dtype=float)
for t in range(1, horizon + 1):
    demand_t = demand1 * (1 + growth) ** (t - 1)
    cash_flow = demand_t * (price - cost)
    npv += cash_flow / (1 + r) ** t

mean_npv = np.mean(npv)
se = np.std(npv, ddof=1) / np.sqrt(N)
ci = (mean_npv - 1.96 * se, mean_npv + 1.96 * se)

print("Mean NPV:", mean_npv)
print("Monte Carlo standard error:", se)
print("95% confidence interval:", ci)
print("Median NPV:", np.median(npv))
````

## Exercise 10(b) (3 points)

Compute the probability of loss, $P(\text{NPV} < 0)$, the 5% VaR, and the 5% CVaR.

````python
probability_of_loss = np.mean(npv < 0)

var_5 = np.percentile(npv, 5)
cvar_5 = np.mean(npv[npv <= var_5])

print("Probability of loss:", probability_of_loss)
print("5% VaR (NPV threshold):", var_5)
print("5% CVaR (average worst 5% NPV):", cvar_5)
````

## Exercise 10(c) (2 points)

Treat the first 1,000 replications as a pilot study. How many replications are needed to estimate the mean NPV within a margin of error of \$5,000 with 95% confidence?

````python
pilot = npv[:1000]

pilot_sd = np.std(pilot, ddof=1)
margin_error = 5000
z = 1.96

required_N = int(np.ceil((z * pilot_sd / margin_error) ** 2))

print("Pilot standard deviation:", pilot_sd)
print("Required number of replications:", required_N)
print("Additional replications needed:", max(0, required_N - 1000))
````

## Exercise 10(d) (3 points)

Perform a sensitivity analysis: compute the squared correlation between each sampled input and the NPV, normalize the values so they sum to 100%, and rank the inputs. Which input is the largest driver of risk, and what action would you recommend?

````python
import pandas as pd

inputs = {
    "Year-1 demand": demand1,
    "Annual demand growth": growth,
    "Price per unit": price,
    "Unit cost": cost
}

squared_corr = {
    name: np.corrcoef(values, npv)[0, 1] ** 2
    for name, values in inputs.items()
}

total = sum(squared_corr.values())

sensitivity = pd.DataFrame({
    "Input": list(squared_corr.keys()),
    "Squared Correlation": list(squared_corr.values())
})

sensitivity["Importance (%)"] = (
    sensitivity["Squared Correlation"] / total * 100
)

sensitivity = sensitivity.sort_values(
    "Importance (%)", ascending=False
).reset_index(drop=True)

sensitivity["Rank"] = np.arange(1, len(sensitivity) + 1)

print(sensitivity.round(4))

print("\nLargest driver of risk:", sensitivity.loc[0, "Input"])
````
